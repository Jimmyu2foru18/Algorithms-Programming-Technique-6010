```python
q1 = np.array([5.6, 3.0, 4.5, 1.5])
t1 = np.array([7.0, 3.2, 4.7, 1.4])
t2 = np.array([6.4, 3.2, 4.5, 1.5])

# q1 → t1
diff1 = t1 - q1
print("q1 → t1: diffs =", diff1, "| sq =", diff1**2,
      "| sum =", round(np.sum(diff1**2), 2),
      "| distance =", round(np.sqrt(np.sum(diff1**2)), 2))

# q1 → t2
diff2 = t2 - q1
print("q1 → t2: diffs =", diff2, "| sq =", diff2**2,
      "| sum =", round(np.sum(diff2**2), 2),
      "| distance =", round(np.sqrt(np.sum(diff2**2)), 2))
```

```python
# Pick a query point from Xtest
query = Xtest.values[0]
print("Query:", query)
# Compute Euclidean distance to every row in Xtest
dists = np.sqrt(np.sum((Xtest.values - query) ** 2, axis=1))
# Find the 3 nearest neighbors using argpartition (faster than full sort)
k = 3
# argpartition gives indices that would sort the array, with the k-th
# smallest element in its sorted position; everything before it is smaller
nearest_idx = np.argpartition(dists, k)[:k]
# Sort those k indices by their actual distance for clean output
nearest_idx = nearest_idx[np.argsort(dists[nearest_idx])]
print("\nNearest neighbor indices:", nearest_idx)
print("Distances:", dists[nearest_idx])
print("Labels:", ytest.values[nearest_idx])
```

```python
# Pick a query point
query = Xtest.values[0]
print("Query:", query)

# Manually compute distance to each row in Xtest using a for loop
for i, row in enumerate(Xtest.values):
    diff = row - query
    sq = diff ** 2
    s = np.sum(sq)
    d = np.sqrt(s)
    print(f"Row {i}: diffs={diff}, sq={sq}, sum={round(s,2)}, dist={round(d,2)}")
```

```python
# Pick a query point
query = Xtest.values[0]
print("Query:", query)

# Compute Euclidean distance to every row in Xtest
dists = np.sqrt(np.sum((Xtest.values - query) ** 2, axis=1))

# argsort returns indices that would sort the array (ascending)
sorted_idx = np.argsort(dists)

# Show the 5 nearest neighbors
k = 5
print("\nTop 5 nearest neighbors:")
for i in sorted_idx[:k]:
    print(f"  Index {i}: dist={round(dists[i], 2)}, label={ytest.values[i]}")
```

```python
# Pick a query point
query = Xtest.values[0]

# Compute distances to every row in Xtest
dists = np.sqrt(np.sum((Xtest.values - query) ** 2, axis=1))

# argsort to rank all rows by distance
sorted_idx = np.argsort(dists)

# Print all rows without truncation
print("All rows sorted by distance (no truncation):")
print(f"{'Rank':<5} {'Index':<8} {'Distance':<10} {'Label':<12} {'Features'}")
print("-" * 70)
for rank, i in enumerate(sorted_idx):
    print(f"{rank:<5} {i:<8} {round(dists[i], 2):<10} {ytest.values[i]:<12} {Xtest.values[i]}")
```

```python
# Partition Xtrain into groups of 3 rows, aggregate each partition (sum)
partition_size = 3
n_partitions = len(Xtrain) // partition_size

# Reshape and sum each partition
partitioned = Xtrain.values[:n_partitions * partition_size].reshape(n_partitions, partition_size, -1)
partition_sums = np.sum(partitioned, axis=1)

print("Total partitions:", n_partitions)
print("Partition sums:", partition_sums)
```

```python
target_idx = 70  # 71st value (0-indexed)
partition_num = target_idx // partition_size
print(f"Index {target_idx} is in partition {partition_num}")
print(f"Partition {partition_num} sum:", partition_sums[partition_num])
print(f"Partition {partition_num} rows:", partitioned[partition_num])
```
