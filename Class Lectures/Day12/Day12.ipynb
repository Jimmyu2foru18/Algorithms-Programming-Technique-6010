## Manually doing KNN on Iris

Target = `species`. Features = sepal/petal length & width.
No `sklearn.neighbors` — we build `ManualKNN` ourselves and pick `k` with an elbow curve.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report


class ManualKNN:
    def __init__(self, k=3):
        self.k = k

    def fit(self, X, y):
        # KNN is lazy: just store the training data
        self.X_train = np.array(X)
        self.y_train = np.array(y)

    def predict(self, X):
        X = np.array(X)
        predictions = [self._predict_single(x) for x in X]
        return np.array(predictions)

    def _predict_single(self, x):
        # 1. Euclidean distance from x to every training row (vectorized)
        distances = np.sqrt(np.sum((self.X_train - x) ** 2, axis=1))

        # 2. argpartition -> indices of the k nearest (unsorted),
        #    then argsort -> order those k by actual distance
        k_indices = np.argpartition(distances, self.k)[:self.k]
        k_indices = k_indices[np.argsort(distances[k_indices])]
        # 3. labels of the k nearest neighbors
        k_nearest_labels = self.y_train[k_indices]
        # 4. majority vote via np.unique (no Counter)
        unique_labels, counts = np.unique(k_nearest_labels, return_counts=True)
        return unique_labels[np.argmax(counts)]
```

### Single-prediction walk-through

We take the first test row, compute distances to all training rows,
pull the 3 nearest with `argpartition`, sort them with `argsort`,
and majority-vote exactly what `_predict_single` does.

```python
# --- Manual walk-through for ONE test point ---
query = Xtest.values[0]
true_label = ytest.values[0]
k = 3
print("Query point (first test row):", query)
print("True label:                   ", true_label)
#distances from query to every training sample
distances = np.sqrt(np.sum((Xtrain.values - query) ** 2, axis=1))
# argpartition for k nearest, then argsort to rank them by distance
k_indices = np.argpartition(distances, k)[:k]
k_indices = k_indices[np.argsort(distances[k_indices])]
# labels of those neighbors
k_nearest_labels = ytrain.values[k_indices]
print(f"\n{k} nearest train indices :", k_indices)
print("Distances                   ", [round(distances[i], 3) for i in k_indices])
print("Neighbor labels             ", k_nearest_labels)
#majority vote via np.unique
unique_labels, counts = np.unique(k_nearest_labels, return_counts=True)
predicted = unique_labels[np.argmax(counts)]
print("Majority vote               ", predicted)
# Sanity check against the class
knn_check = ManualKNN(k=k)
knn_check.fit(Xtrain, ytrain)
print("Class predict() on point    ", knn_check.predict([query])[0])
```

### Elbow curve For the accuracy vs. k

Sweep `k` from 1 to 30, score each `ManualKNN` on the held-out test set,
and plot accuracy. The "elbow" is where accuracy stops improving meaningfully.

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

X = irisDF[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
Xs = StandardScaler().fit_transform(X)

inertias = []
for k in range(1, 11):
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    km.fit(Xs)
    inertias.append(km.inertia_)

plt.plot(range(1, 11), inertias, marker='o')
plt.xlabel('k'); plt.ylabel('Inertia'); plt.title('Elbow Curve (Iris)')
plt.show()
```

```python
k_values = list(range(1, 31))
accuracies = []
for k in k_values:
    knn = ManualKNN(k=k)
    knn.fit(Xtrain, ytrain)
    preds = knn.predict(Xtest)
    accuracies.append(accuracy_score(ytest, preds))
# Elbow curve
plt.figure(figsize=(8, 5))
plt.plot(k_values, accuracies, marker='o', linestyle='-', color='steelblue')
plt.xlabel('k (number of neighbors)')
plt.ylabel('Test accuracy')
plt.title('KNN Elbow Curve: Accuracy vs. k (Iris dataset)')
plt.grid(True, alpha=0.3)
plt.xticks(k_values)
plt.ylim(0, 1)
plt.show()
# Print table
for k, acc in zip(k_values, accuracies):
    print(f"k={k:2d} -> accuracy = {acc:.3f}")
best_k = k_values[int(np.argmax(accuracies))]
print(f"\nBest k = {best_k} with test accuracy {max(accuracies):.3f}")
```

## KNN on Iris: Accuracy vs. k (not a true elbow curve)

A real **elbow curve** plots KMeans **inertia** (within-cluster sum of squares) vs. number of clusters — it is used for clustering, not classification.

Here we sweep `k = 1..30` for **KNN classification** and plot **test accuracy vs. k**. This curve typically plateaus rather than forming a sharp elbow; pick `k` at the peak or where accuracy stops improving.

Important: KNN is distance-based, so features are scaled with `StandardScaler` first. Without scaling, large-range features (e.g. petal length) dominate the distance and the curve looks flat or misleading.


```python
# KNN accuracy vs k on the real Iris data (not a true elbow curve; elbow curves plot KMeans inertia)
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

X = irisDF[["sepal_length", "sepal_width", "petal_length", "petal_width"]]
y = irisDF["species"]

Xtrain, Xtest, ytrain, ytest = train_test_split(X, y, test_size=0.20, random_state=42)

# Scale features -- KNN is distance-based, unscaled features distort distances
scaler = StandardScaler()
Xtrain_s = scaler.fit_transform(Xtrain)
Xtest_s  = scaler.transform(Xtest)

class ManualKNN:
    def __init__(self, k=3):
        self.k = k
    def fit(self, X, y):
        self.X_train = np.asarray(X)
        self.y_train = np.asarray(y)
    def predict(self, X):
        X = np.asarray(X)
        preds = []
        for x in X:
            dists = np.sqrt(np.sum((self.X_train - x) ** 2, axis=1))
            nearest = np.argsort(dists)[: self.k]
            votes = self.y_train[nearest]
            preds.append(max(set(votes), key=list(votes).count))
        return np.array(preds)

k_values = list(range(1, 31))
accuracies = []
for k in k_values:
    knn = ManualKNN(k=k)
    knn.fit(Xtrain_s, ytrain)
    preds = knn.predict(Xtest_s)
    accuracies.append(accuracy_score(ytest, preds))

plt.figure(figsize=(8, 5))
plt.plot(k_values, accuracies, marker="o", linestyle="-", color="steelblue")
plt.xlabel("k (number of neighbors)")
plt.ylabel("Test accuracy")
plt.title("KNN Accuracy vs. k (Iris dataset)")
plt.grid(True, alpha=0.3)
plt.xticks(k_values)
plt.ylim(0, 1)
plt.show()

for k, acc in zip(k_values, accuracies):
    print(f"k={k:2d} -> accuracy = {acc:.3f}")
best_k = k_values[int(np.argmax(accuracies))]
print(f"\nBest k = {best_k} with test accuracy {max(accuracies):.3f}")
```

### Train with chosen k and evaluate on test set

```python
# Final model with the chosen k
knn = ManualKNN(k=best_k)
knn.fit(Xtrain_s, ytrain)
predictions = knn.predict(Xtest_s)
print(f"Final k = {best_k}")
print("\nFirst 10 predictions:", predictions[:10])
print("First 10 true labels:", ytest.values[:10])

print(f"\nTest accuracy : {accuracy_score(ytest, predictions):.3f}")
print("\nConfusion matrix:\n", confusion_matrix(ytest, predictions))
print("\nClassification report:\n", classification_report(ytest, predictions))
```
