```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
df = pd.read_csv('olympics.csv')
df.head()
```

```python
df.info()
```

```python
# Rename columns
df.columns = ['Country', 'Summer Games', 'Summer Gold', 'Summer Silver', 'Summer Bronze', 'Summer Total',
              'Winter Games', 'Winter Gold', 'Winter Silver', 'Winter Bronze', 'Winter Total',
              'Total Games', 'Total Gold', 'Total Silver', 'Total Bronze', 'Combined Total']

# Drop rows with any NaN
df = df.dropna()

# Convert each column to int
df['Summer Games'] = pd.to_numeric(df['Summer Games'], errors='coerce').astype('int64')
df['Summer Gold'] = pd.to_numeric(df['Summer Gold'], errors='coerce').astype('int64')
df['Summer Silver'] = pd.to_numeric(df['Summer Silver'], errors='coerce').astype('int64')
df['Summer Bronze'] = pd.to_numeric(df['Summer Bronze'], errors='coerce').astype('int64')
df['Summer Total'] = pd.to_numeric(df['Summer Total'], errors='coerce').astype('int64')
df['Winter Games'] = pd.to_numeric(df['Winter Games'], errors='coerce').astype('int64')
df['Winter Gold'] = pd.to_numeric(df['Winter Gold'], errors='coerce').astype('int64')
df['Winter Silver'] = pd.to_numeric(df['Winter Silver'], errors='coerce').astype('int64')
df['Winter Bronze'] = pd.to_numeric(df['Winter Bronze'], errors='coerce').astype('int64')
df['Winter Total'] = pd.to_numeric(df['Winter Total'], errors='coerce').astype('int64')
df['Total Games'] = pd.to_numeric(df['Total Games'], errors='coerce').astype('int64')
df['Total Gold'] = pd.to_numeric(df['Total Gold'], errors='coerce').astype('int64')
df['Total Silver'] = pd.to_numeric(df['Total Silver'], errors='coerce').astype('int64')
df['Total Bronze'] = pd.to_numeric(df['Total Bronze'], errors='coerce').astype('int64')
df['Combined Total'] = pd.to_numeric(df['Combined Total'], errors='coerce').astype('int64')

df.head()
```

```python
df.set_index("Country", inplace=True)
```

```python
# which country has the most gold in the winter and bronze in the summer
# and then
# most metals per apearance make a column that has the most metals per apearance
# and then 
# if gold is 3, silver is 2, and bronz is 1 give a total number of points for each.
```

```python
# Which country has the most gold in winter
most_winter_gold_country = df['Winter Gold'].idxmax()
most_winter_gold_value = df['Winter Gold'].max()

print(f"Country with most Winter Gold: {most_winter_gold_country} ({most_winter_gold_value} gold medals)")

# Which country has the most bronze in summer
most_summer_bronze_country = df['Summer Bronze'].idxmax()
most_summer_bronze_value = df['Summer Bronze'].max()
print(f"Country with most Summer Bronze: {most_summer_bronze_country} ({most_summer_bronze_value} bronze medals)")
```

```python
# Most metals per appearance (Total medals / Total Games)
df['Metals per Appearance'] = df['Combined Total'] / df['Total Games']
most_metals_per_appearance = df['Metals per Appearance'].idxmax()
most_metals_value = df['Metals per Appearance'].max()
print(f"Country with most metals per appearance: {most_metals_per_appearance} ({most_metals_value:.2f} medals per game)")
df.head()
```

```python
# Points system: Gold=3, Silver=2, Bronze=1
df['Total Points'] = (df['Total Gold'] * 3) + (df['Total Silver'] * 2) + (df['Total Bronze'] * 1)
df['Summer Points'] = (df['Summer Gold'] * 3) + (df['Summer Silver'] * 2) + (df['Summer Bronze'] * 1)
df['Winter Points'] = (df['Winter Gold'] * 3) + (df['Winter Silver'] * 2) + (df['Winter Bronze'] * 1)

# Show top countries by total points
top_points = df.sort_values('Total Points', ascending=False).head(10)
print(top_points[['Total Points', 'Summer Points', 'Winter Points']])
```
