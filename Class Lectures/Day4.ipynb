## Exercise 1
Create a list of three of your favorite foods. Save it in a variable called `foods`.

```python
# A list literal is created with square brackets, with items separated by commas.
# Strings can use single or double quotes; here we use single quotes for consistency.
foods = ['Fish and Chips', 'African Vegetables', 'Chilli']

# print() displays the value so we can confirm the list was created correctly.
print(foods)
```

## Exercise 2
From your list of favorite foods, select the second item.

```python
# Python lists are zero-indexed, so the FIRST item is at index 0
# and the SECOND item is at index 1.
second_food = foods[1]
print(second_food)
```

## Exercise 3
Using a slice, create a **reversed** copy of your list of favorite foods.

```python
# Slice syntax is list[start:stop:step].
# Leaving start and stop empty means "the whole list", and a step of -1
# walks the list backwards, one item at a time -- producing a reversed COPY.
# NOTE: This does NOT modify the original `foods` list.
reversed_foods = foods[::-1]
print(reversed_foods)

# Confirm the original list is untouched:
print('original foods list unchanged:', foods)
```

## Exercise 4
`"my name is john".upper()` returns the string `"MY NAME IS JOHN"`. Given that fact, create an uppercase list of your favorite foods using a `for` loop.

```python
# The classic "accumulator pattern":
# 1. Start with an empty list.
# 2. Loop over each item in the source list.
# 3. Transform the item and append the result to the new list.
uppercase_foods_loop = []
for food in foods:
    uppercase_foods_loop.append(food.upper())

print(uppercase_foods_loop)
```

## Exercise 5
Create a second uppercase list of your favorite foods, using a **list comprehension**.

```python
# A list comprehension does the same accumulator pattern as Exercise 4,
# but in a single, more "Pythonic" expression:
#   [expression for item in iterable]
uppercase_foods_comprehension = [food.upper() for food in foods]
print(uppercase_foods_comprehension)

# Sanity check: both approaches should give identical results.
print('loop result == comprehension result:', uppercase_foods_loop == uppercase_foods_comprehension)
```
