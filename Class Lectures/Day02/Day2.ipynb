## Organize List into Non-Smokers & Smokers

```python
presidents_data.sort(key=lambda x: x[3])
```

```python
presidents_data
```

## Filter for Smokers

```python
def checkSmoking(item):
    return item[3]
```

```python
presidentSmokers = list(filter(checkSmoking, presidents_data))
```

```python
presidentSmokers
```

### Or --- Just by Name

```python
smokers = [p[0] for p in presidents_data if p[3]]
print(smokers)
```

## Filter for Non-Smokers

```python
def checkNonSmoking(item):
    return not item[3]
```

```python
presidentNonSmokers = list(filter(checkNonSmoking, presidents_data))
```

```python
presidentNonSmokers
```

### Or --- Just by Name

```python
non_smokers = [p[0] for p in presidents_data if not p[3]]
print(non_smokers)
```

## Sort Presidents by height as function

```python
def sort_presidents_by_height(data, reverse=False):
    return sorted(data, key=lambda x: x[1], reverse=reverse)
sorted_presidents = sort_presidents_by_height(presidents_data)
for president in sorted_presidents:
    print(f"{president[0]}: {president[1]}")
```

## List comprehension for by height

```python
by_height = sorted(presidents_data, key=lambda x: x[1])
print(by_height)
```
