* d1 = {'a': 100, 'b': 200, 'c':300}

* Sample output: Counter({'a': 400, 'b': 400, 'd': 400, 'c': 300})

#### 3. Write a Python program to print all unique values in a dictionary
* Sample Data : [{"V":"S001"}, {"V": "S002"}, {"VI": "S001"}, {"VI": "S005"}, {"VII":"S005"}, {"V":"S009"},{"VIII":"S007"}]
* Expected Output : Unique Values: {'S005', 'S002', 'S007', 'S001', 'S009'}

```python
sample_data = [
    {"V": "S001"},
    {"V": "S002"},
    {"VI": "S001"},
    {"VI": "S005"},
    {"VII": "S005"},
    {"V": "S009"},
    {"VIII": "S007"},
]

unique_values = set()
for d in sample_data:
  for val in d.values():
    unique_values.add(val)

print("Unique Values:", unique_values)
```

#### 4. Write a Python program to create and display all combinations of letters, selecting each letter from a different key in a dictionary.
* Sample data : {'1':['a','b'], '2':['c','d']}
###### Expected Output: 
ac
ad
bc
bd

```python
def generate_combinations(data):
  keys = list(data.keys())
  def helper(index, current):
    if index == len(keys):
      print("".join(current))
      return
    for char in data[keys[index]]:
      helper(index + 1, current + [char])
  helper(0, [])
sample_data = {'1': ['a', 'b'], '2': ['c', 'd']}  #[cite: 1]
generate_combinations(sample_data)
```

#### 5. Write a Python program to create a dictionary from a string.
* Note: Track the count of the letters from the string.
* Sample string : 'w3resource'
* Expected output: {'3': 1, 's': 1, 'r': 2, 'u': 1, 'w': 1, 'c': 1, 'e': 2, 'o': 1}

```python
sample_string = input("Please input a string to be made into a dictionary of letters and Frequencies: ")

chars = [c for c in sample_string]
result = {char: len([c for c in chars if c == char]) for char in chars}

print(result)
```
