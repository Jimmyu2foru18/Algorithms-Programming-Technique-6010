### 6-5. Rivers

Make a dictionary containing three major rivers and the country each river runs through. One key-value pair might be `'nile': 'egypt'`.

- Use a loop to print a sentence about each river, such as *The Nile runs through Egypt.*
- Use a loop to print the name of each river included in the dictionary.
- Use a loop to print the name of each country included in the dictionary.

```python
rivers = {
    "nile": "egypt",
    "amazon": "brazil",
    "yangtze": "china"
}

# Sentence about each river
for river, country in rivers.items():
    print(f"The {river.title()} runs through {country.title()}.")

# Names of rivers
print("\nRivers:")
for river in rivers.keys():
    print(river.title())

# Names of countries
print("\nCountries:")
for country in rivers.values():
    print(country.title())
```

Start with the program you wrote for Exercise 6-1. Make two new dictionaries representing different people and store all three dictionaries in a list called `people`. Loop through your list of people. As you loop through the list, print everything you know about each person.

### 6-9. Favorite Places

### 6-11. Cities

Make a dictionary called `cities`. Use the names of three cities as keys in your dictionary. Create a dictionary of information about each city and include the country that the city is in, its approximate population, and one fact about that city. The keys for each city's dictionary should be something like `country`, `population`, and `fact`. Print the name of each city and all of the information you have stored about it.

```python
cities = {
    "rome": {
        "country": "italy",
        "population": 2_873_000,
        "fact": "Rome is home to the Colosseum, an ancient amphitheater."
    },
    "athens": {
        "country": "greece",
        "population": 664_000,
        "fact": "Athens is considered the cradle of Western civilization."
    },
    "miami": {
        "country": "united states",
        "population": 442_000,
        "fact": "Miami is known for its vibrant Latin American culture."
    }
}

for city, info in cities.items():
    print(f"\nCity: {city.title()}")
    for key, value in info.items():
        print(f"  {key}: {value}")
```

### 6-12. Extensions

We're now working with examples that are complex enough that they can be extended in any number of ways. Use one of the example programs from this chapter, and extend it by adding new keys and values, changing the context of the program, or improving the formatting of the output.

```python
# Extension of 6-11 Cities with better formatting and additional data
cities = {
    "rome": {
        "country": "italy",
        "population": 2_873_000,
        "fact": "Rome is home to the Colosseum, an ancient amphitheater.",
        "language": "Italian",
        "currency": "EUR"
    },
    "athens": {
        "country": "greece",
        "population": 664_000,
        "fact": "Athens is considered the cradle of Western civilization.",
        "language": "Greek",
        "currency": "EUR"
    },
    "miami": {
        "country": "united states",
        "population": 442_000,
        "fact": "Miami is known for its vibrant Latin American culture.",
        "language": "English",
        "currency": "USD"
    }
}

for city, info in cities.items():
    print(f"\n{'='*40}")
    print(f"City: {city.title()}")
    print(f"{'='*40}")
    for key, value in info.items():
        print(f"{key.title()}: {value}")
```
