### 10-6. Addition

One common problem when prompting for numerical input occurs when people provide text instead of numbers. When you try to convert the input to an `int`, you'll get a `ValueError`. Write a program that prompts for two numbers. Add them together and print the result. Catch the `ValueError` if either input value is not a number, and print a friendly error message. Test your program by entering two numbers and then by entering some text instead of a number.

````python
# Test 1: Valid numbers
try:
    num1 = 20  # int(input("Enter first number: "))
    num2 = 22  # int(input("Enter second number: "))
    result = num1 + num2
    print(f"Result: {result}")
except ValueError:
    print("Please enter valid numbers!")

# Test 2: Invalid input
try:
    num1 = "hello"  # int(input("Enter first number: "))
    num2 = 10  # int(input("Enter second number: "))
    result = num1 + num2
    print(f"Result: {result}")
except ValueError:
    print("Oops! That's not a valid number. Please try again.")
````

### 10-7. Addition Calculator

Wrap your code from Exercise 10-6 in a `while` loop so the user can continue entering numbers even if they make a mistake and enter text instead of a number.

````python
print("Addition Calculator")
print("Enter 'q' at any time to quit.\n")

while True:
    num1_input = input("Enter first number (or 'q' to quit): ")
    if num1_input.lower() == "q":
        break

    num2_input = input("Enter second number (or 'q' to quit): ")
    if num2_input.lower() == "q":
        break

    try:
        num1 = int(num1_input)
        num2 = int(num2_input)
    except ValueError:
        print("Oops! Please enter valid numbers.\n")
        continue

    result = num1 + num2
    print(f"Result: {result}\n")
````

### 10-8. Cats and Dogs

Make two files, `cats.txt` and `dogs.txt`. Store at least three names of cats in the first file and three names of dogs in the second file. Write a program that tries to read these files and print the contents of the file to the screen. Wrap your code in a `try-except` block to catch the `FileNotFound` error, and print a friendly message if a file is missing.

````python
# Create sample files
with open("cats.txt", "w") as file:
    file.write("Shadow\nGhost\nWhiskers\n")

with open("dogs.txt", "w") as file:
    file.write("Rocket\nBuddy\nRocky\n")

# Read and print files
files = ["cats.txt", "dogs.txt", "missing.txt"]

for filename in files:
    try:
        with open(filename) as file:
            print(f"\n{filename}:")
            print(file.read())
    except FileNotFoundError:
        print(f"Sorry, the file {filename} is missing.")
````

### 10-10. Common Words

Visit [Project Gutenberg](https://gutenberg.org/) and find a few texts you'd like to analyze. Download the text files for these works, or copy the raw text from your browser into a text file on your computer.

You can use the `count()` method to find out how many times a word or phrase appears in a string. For example, the following code counts the number of times `'row'` appears in a string:

```python
>>> line = "Row, row, row your boat"
>>> line.count('row')
2
>>> line.lower().count('row')

```

Notice that converting the string to lowercase using `lower()` catches all appearances of the word you're looking for, regardless of how it's formatted.

Write a program that reads the files you found at Project Gutenberg and determines how many times the word `'the'` appears in each text. This will be an approximation because it will also count words such as `'then'` and `'there'`. Try counting `'the '`, with a space in the string, and see how much lower your count is.

**Note:** For demonstration, we'll use a sample text string instead of downloading from Project Gutenberg.

````python
# Sample text for demonstration
sample_text = """
The quick brown fox jumps over the lazy dog.
The dog was not amused by the fox's behavior.
Then the fox ran away into the forest.
There the fox found a new friend.
"""

lower_text = sample_text.lower()

# Count 'the' (approximate)
count_the = lower_text.count("the")
print(f"Occurrences of 'the': {count_the}")

# Count 'the ' (more accurate)
count_the_space = lower_text.count("the ")
print(f"Occurrences of 'the ': {count_the_space}")
````

### 10-11. Favorite Number

Write a program that prompts for the user's favorite number. Use `json.dump()` to store this number in a file. Write a separate program that reads in this value and prints the message, *"I know your favorite number! It's _____."*

````python
import json

# Program 1: Store favorite number
number = 20  # int(input("What is your favorite number? "))

with open("favorite_number.json", "w") as file:
    json.dump(number, file)

# Program 2: Read favorite number
with open("favorite_number.json") as file:
    favorite_number = json.load(file)

print(f"I know your favorite number! It's {favorite_number}.")
````

### 10-12. Favorite Number Remembered

Combine the two programs from Exercise 10-11 into one file. If the number is already stored, report the favorite number to the user. If not, prompt for the user's favorite number and store it in a file. Run the program twice to see that it works.

````python
import json
import os

filename = "favorite_number.json"

# Check if file exists
if os.path.exists(filename):
    with open(filename) as file:
        favorite_number = json.load(file)
    print(f"I know your favorite number! It's {favorite_number}.")
else:
    number = int(input("What is your favorite number? "))
    with open(filename, "w") as file:
        json.dump(number, file)
    print(f"I'll remember your favorite number: {number}.")

# Simulate second run
print("\nSecond run:")
if os.path.exists(filename):
    with open(filename) as file:
        favorite_number = json.load(file)
    print(f"I know your favorite number! It's {favorite_number}.")
````

### 10-13. Verify User

The final listing for `remember_me.py` assumes either that the user has already entered their username or that the program is running for the first time. We should modify it in case the current user is not the person who last used the program.

Before printing a welcome back message in `greet_user()`, ask the user if this is the correct username. If it's not, call `get_new_username()` to get the correct username.

````python
import json

def get_stored_username():
    """Get stored username if available."""
    filename = "username.json"
    try:
        with open(filename) as file:
            username = json.load(file)
    except FileNotFoundError:
        return None
    else:
        return username


def get_new_username():
    """Prompt for a new username."""
    username = input("What is your name? ")
    filename = "username.json"
    with open(filename, "w") as file:
        json.dump(username, file)
    return username


def greet_user():
    """Greet the user by name."""
    username = get_stored_username()
    if username:
        correct = input(f"Is '{username}' your username? (yes/no) ")
        if correct.lower() == "yes":
            print(f"Welcome back, {username}!")
        else:
            username = get_new_username()
            print(f"We'll remember you when you come back, {username}!")
    else:
        username = get_new_username()
        print(f"We'll remember you when you come back, {username}!")


# Demonstrate the logic (simulating stored username exists)
# For first run demonstration, we'll create the file
with open("username.json", "w") as file:
    json.dump("James McGuigan", file)

# Now simulate the greet_user flow
username = get_stored_username()
print(f"Stored username: {username}")

# Simulate user confirming it's correct
# In real use: correct = input(f"Is '{username}' your username? (yes/no) ")
correct = "yes"  # Simulated input
if correct.lower() == "yes":
    print(f"Welcome back, {username}!")
else:
    username = get_new_username()
    print(f"We'll remember you when you come back, {username}!")
````
