## 3-7. Shrinking Guest List

Remove guests until only two remain.

```python
# Full guest list
guests = ["Marie Curie", "James Gosling", "Nikola Tesla", "Ada Lovelace", "Linus Torvalds", "Isaac Newton"]

# Only two can come
print("Sorry, I can only invite two people for dinner.\n")

# Remove guests
while len(guests) > 2:
	removed_guest = guests.pop()
	print(f"Sorry {removed_guest}, I can't invite you to dinner.")

# Remaining guests still invited
print(f"\n{guests[0]}, you're still invited!")
print(f"{guests[1]}, you're still invited!")

# Clear list
del guests[0]
del guests[0]

# Verify empty
print(f"\nFinal guest list: {guests}")
```

## 3-8. Seeing the World

Sort and reverse places list.

```python
# Places to visit
places = ["Tokyo", "Paris", "Greece", "Sydney", "Cairo"]

# Original order
print("Original order:")
print(places)

# Alphabetical
print("\nAlphabetical order (sorted()):")
print(sorted(places))

# Original again
print("\nOriginal order again:")
print(places)

# Reverse alphabetical
print("\nReverse alphabetical order (sorted()):")
print(sorted(places, reverse=True))

# Original again
print("\nOriginal order again:")
print(places)

# Reverse list
print("\nReversed order (reverse()):")
places.reverse()
print(places)

# Back to original
print("\nReversed again (back to original):")
places.reverse()
print(places)

# Sort alphabetically
print("\nSorted alphabetically (sort()):")
places.sort()
print(places)

# Sort reverse
print("\nSorted reverse alphabetically (sort()):")
places.sort(reverse=True)
print(places)
```

## 3-10. Every Function

Use all list functions.

```python
# Countries list
countries = ["Japan", "France", "Brazil", "Australia", "Egypt"]
print("Original list:")
print(countries)

# Access elements
print(f"\nFirst country: {countries[0]}")
print(f"Last country: {countries[-1]}")

# Modify element
countries[0] = "Canada"
print(f"\nAfter modification: {countries}")

# Append
countries.append("Japan")
print(f"After append: {countries}")

# Insert
countries.insert(2, "Germany")
print(f"After insert: {countries}")

# Remove by value
countries.remove("Brazil")
print(f"After remove: {countries}")

# Pop
popped = countries.pop()
print(f"Popped: {popped}")
print(f"After pop: {countries}")

# Sort temporarily
print(f"\nSorted (temporary): {sorted(countries)}")
print(f"Original still: {countries}")

# Sort permanently
countries.sort()
print(f"\nAfter sort(): {countries}")

# Reverse
countries.reverse()
print(f"After reverse(): {countries}")

# Length
print(f"\nLength: {len(countries)}")

# Delete by index
del countries[0]
print(f"After del: {countries}")
```

## 3-11. Intentional Error

Learn about IndexError by causing one on purpose.

An IndexError happens when you try to access an index that doesn't exist in a list.
For example, if a list has 3 items (indices 0, 1, 2), trying to access index 5 will cause an error.

Steps:
1. Create a short list
2. Try to access an index that is out of range
3. See the error message
4. Fix the error by using a valid index

```python
# Short list
fruits = ["apple", "banana", "cherry"]

# Valid indices: 0, 1, 2
# This will cause an IndexError:
# print(fruits[5])  # IndexError: list index out of range

# Fixed - use valid index:
print(fruits[2])  # Prints 'cherry'
print(f"Valid indices are 0 to {len(fruits) - 1}")
print(f"List has {len(fruits)} items")
```
