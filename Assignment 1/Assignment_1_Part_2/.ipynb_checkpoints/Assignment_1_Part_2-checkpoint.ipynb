## 1A) Remove duplicate values in one line

```
p = [1,2,2,9,3,4,3,3,2,5,4,3,3,34,2,3,3,3,2,1]
list(set(p))
```

A set is an unordered collection of unique elements, so passing the list to the `set()` constructor automatically discards all repeat values. As shown in the text, `set([2, 2, 2, 1, 3, 3])` evaluates to `{1, 2, 3}`. Wrapping the result in `list()` converts it back to a list if needed.

## 1B) How many repeat values were removed?

```
p = [1,2,2,9,3,4,3,3,2,5,4,3,3,34,2,3,3,3,2,1]
len(p) - len(set(p))
```

The `len()` function returns the number of elements in a collection. By subtracting the length of the deduplicated set from the original list length, you directly count how many repeat values were removed.

## 2) Slice `myList`

```
myList = ["a", "o", "l", "d", "w", "e", "s", "t", "p", "q"]
myList[1:8]
```

Slice notation uses `start:stop` to select elements, where the start index is included and the stop index is excluded. Therefore `myList[1:8]` returns the elements at indices 1 through 7, giving `["o","l","d","w","e","s","t"]`.

```
myList = ["a", "o", "l", "d", "w", "e", "s", "t", "p", "q"]
myList[0:9:2]
```

To pick non-consecutive elements, add a step value after a second colon. `myList[0:9:2]` (or simply `myList[::2]`) selects every other element starting from index 0, producing `["a", "l", "w", "s", "p"]`.

## 3) List comprehension to square all contents

```
numbers = [1, 2, 3, 4, 5]
[x**2 for x in numbers]
```

List comprehensions take the basic form `[expr for value in collection if condition]`, letting you replace a multi-line for loop with a single expression. Using `[x**2 for x in numbers]` applies the squaring operation to every element in one concise line.

## 4) First 100 multiples of 5 in descending order

```
list(range(500, 0, -5))
```

The `range()` function accepts a start, end, and step, where the step may be negative to count downward. Using `range(500, 0, -5)` generates the first 100 multiples of 5 in descending order, because it starts at 500 and steps by -5 until it reaches 0.

## 5) Can a list have elements of different datatypes?

```
myList = ["hello", 5, "yes", 19.2]
print(myList)
```

Yes, a Python list can contain elements of different datatypes. The text demonstrates this with examples like `[2, 3, 7, None]` and `[4, None, "foo"]`, showing that lists are mutable and can hold any mix of objects without requiring uniform types.
