####
- **Series**: One dimensional labeled array.
- **DataFrame**: Two dimensional table structure.

- Inspection methods:
    - df.head(n) views top rows.
    - df.tail(n) views bottom rows.
    - df.sample(n) gets random rows.
    - df.info() checks types, nulls, memory.
    - df.describe() shows statistical summaries.
    - df.shape checks dimensions.
    - df.size counts total items.

```python
import pandas as pd
import numpy as np

# one dimensional array
scores = pd.Series([88, 92, 75, 61, 95],
                   index=['alice', 'bob', 'carol', 'dave', 'erin'],
                   name='score')
scores

# two dimensional table structure
df = pd.DataFrame({
    'product': ['Laptop', 'Phone', 'Tablet', 'Monitor', 'Keyboard'],
    'category': ['Electronics', 'Electronics', 'Electronics', 'Electronics', 'Accessories'],
    'price': [1200, 800, 450, 300, 60],
    'units_sold': [12, 40, 25, 18, 100]
})
df

# methods
df.head(3)     # top rows
df.tail(2)     # bottom rows
df.sample(2)   # random rows
df.info()      # types, nulls, memory
df.describe()  # statistical summaries
df.shape       # dimensions
df.size        # total number of items
# or this to decribe type
scores.describe()
```

#### Input and Output
- **CSV files**: pd.read_csv('path.csv')
- **Excel files**: pd.read_excel('path.xlsx', sheet_name='Sheet1')
- **Parquet or Feather files**: pd.read_parquet('path.parquet') for speed.
- **Exporting data**: df.to_csv('path.csv', index=False)

#### Data Access and Indexing
- **Column access**: df['col_name'] or df.col_name
- **Label indexing**: df.loc[]
- **Integer position indexing**: df.iloc[]
- **Fast scalar lookup**: df.at[] and df.iat[]
- **Sorting**: df.sort_values(by='col', ascending=False)

```python
# Column access
df['product']
df.price # attribute

# Label indexing
df.loc[0:2, ['product', 'price']]

# Integer indexing
df.iloc[0:3, 1:3]

# Fast scalar lookup
df.at[0, 'product'] # by label
df.iat[0, 1] # by position

# Sorting
df.sort_values(by='price', ascending=False)
```

#### Filtering Data
- **Boolean indexing**: df[df['col'] > value]
- **Multiple conditions**: use parentheses and & or |
- **String operations**: df['col'].str.contains('pattern', case=False)
- **String matching**: df['col'].str.startswith() or df['col'].isin([...])

```python
# Boolean indexing
df[df['price'] > 400]
df[(df['price'] > 400) & (df['units_sold'] > 20)]
df[(df['category'] == 'Accessories') | (df['price'] > 1000)]

# String 
df[df['product'].str.contains('phone', case=False)]
df[df['product'].str.startswith('T')]
df[df['category'].isin(['Electronics', 'Accessories'])]
```

#### Column Manipulation
- **Adding columns**: df['new_col'] = df['col1'] * df['col2']
- **Conditional assignment**: np.where(condition, true_val, false_val)
- **Renaming columns**: df.rename(columns={'old': 'new'}, inplace=True)
- **Dropping columns**: df.drop(columns=['col_name'], inplace=True)
- **Copying data**: df.copy()

```python
# Adding columns
df['revenue'] = df['price'] * df['units_sold']
df['tier'] = np.where(df['revenue'] > 15000, 'High', 'Standard')
df.rename(columns={'tier': 'value_tier'}, inplace=True)
data = df.copy()
df.drop(columns=['value_tier'], inplace=True)
data
```

#### Datetime Operations
- **Conversion**: pd.to_datetime()
- **Extraction**: df['date_col'].dt.year, df['date_col'].dt.month, df['date_col'].dt.day_of_week

```python
orders = pd.DataFrame({
    'order_id': [1, 2, 3, 4],
    'date': ['2024-01-15', '2024-01-29', '2024-02-03', '2024-02-18'],
    'sales': [120.50, 80.00, 200.00, 65.25]
})

# Conversion
orders['date'] = pd.to_datetime(orders['date'])
orders['date'].dtype

# Extraction
orders['year'] = orders['date'].dt.year
orders['month'] = orders['date'].dt.month
orders['day_of_week'] = orders['date'].dt.day_of_week
orders
```

#### Custom Functions
- **Apply method**: df.apply(lambda x: ..., axis=1)

```python
# run a function on every row
df.apply(lambda row: row['price'] * row['units_sold'], axis=1)

# using a named function
def revenue_band(revenue):
    return 'high' if revenue > 15000 else 'low'

df['revenue_band'] = df.apply(lambda row: revenue_band(row['revenue']), axis=1)
df
```

#### Merging and Concatenating
- **Merging datasets**: pd.merge(df1, df2, left_on='a', right_on='b', how='left')
- **Concatenating datasets**: pd.concat([df1, df2])

```python
targets = pd.DataFrame({
    'item': ['Laptop', 'Phone', 'Tablet', 'Monitor', 'Keyboard'],
    'sales_target': [150, 400, 300, 200, 90]
})

# Merging data
pd.merge(df, targets, left_on='product', right_on='item', how='left')

# Concatenation data
pd.concat([df.iloc[:2], df.iloc[2:]])
```

#### Handling Missing Values
- **Detection**: df.isna().sum()
- **Imputation**: df.fillna(value) or df.interpolate()
- **Dropping nulls**: df.dropna(subset=['col'])

```python
missing = df.copy()
missing.loc[1, 'price'] = np.nan
missing.loc[3, 'units_sold'] = np.nan

# Detection
missing.isna().sum()

# Imputation
missing['price'].fillna(missing['price'].mean())
missing['units_sold'] = missing['units_sold'].fillna(0)
missing['price'] = missing['price'].interpolate()
missing

# Dropping nulls
missing.dropna(subset=['price'])
```

#### Aggregation and Grouping
- **Value counts**: df['col'].value_counts()
- **Group by**: df.groupby('col').agg({'col1': 'sum'})
- **Pivot tables**: df.pivot_table(index='col1', columns='col2', values='col3', aggfunc='sum')

```python
# Value counts
df['category'].value_counts()

# Group by
df.groupby('category').agg({'units_sold': 'sum', 'revenue': 'mean'})

# Pivot tables
df.pivot_table(index='category', columns='product', values='revenue', aggfunc='sum')
regions = pd.DataFrame({
    'product': ['Laptop', 'Phone', 'Tablet', 'Monitor', 'Keyboard'],
    'region': ['North', 'South', 'North', 'South', 'North']
})
with_region = pd.merge(df, regions, on='product', how='inner')
with_region.pivot_table(index='category', columns='region', values='revenue', aggfunc='sum')
```

- **Shifting**: df['col'].shift(n)
- **Rolling windows**: df['col'].rolling(window=n).sum()
- **Ranking**: df['col'].rank(ascending=False)
- **PyArrow backend**: modern performance optimization for strings

```python
import importlib.util
import pandas as pd
import pyarrow

daily = orders.sort_values('date').reset_index(drop=True)
daily['prev_sales'] = daily['sales'].shift(1)
daily['sales_change'] = daily['sales'] - daily['sales'].shift(1)
daily['rolling_sum'] = daily['sales'].rolling(window=2).sum()
daily['rolling_mean'] = daily['sales'].rolling(window=2).mean()
daily['sales_rank'] = daily['sales'].rank(ascending=False)
print(daily.dtypes)
```

## Analysis Questions

- **Question**: Which category generated the highest total revenue?


```python
# Group data
revenue_by_category = df.groupby('category')['revenue'].sum()
revenue_by_category

# getting the highest total revenue
revenue_by_category.idxmax()
revenue_by_category.max()
```

- **Answer**: Group data by category and aggregate total sales. Example: df.groupby('category')['revenue'].sum()

---

- **Question**: How do sales in January compare to February?

```python
# Parse dates
monthly_sales = orders.groupby(orders['date'].dt.month)['sales'].sum()
monthly_sales

january = monthly_sales.loc[1]
february = monthly_sales.loc[2]

print(f"January total:   {january:.2f}")
print(f"February total:  {february:.2f}")
print(f"Difference:      {february - january:+.2f}")
```

- **Answer**: Parse dates, filter by month, and compare sums or use shift. Example: df.groupby(df['date'].dt.month)['sales'].sum()

---

- **Question**: Are there missing values or anomalies in specific columns?

```python
check = orders.copy()
check.loc[1, 'sales'] = np.nan
# Count missing entries
check.isna().sum()
check['sales'].describe()

# Values more than 2 standard deviations
threshold = check['sales'].mean() + 2 * check['sales'].std()
check[check['sales'] > threshold]
```

- **Answer**: Count missing entries and review summary statistics. Example: df.isna().sum() and df.describe()

- **Question**: How do two separate datasets relate to each other?


```python
# second dataset
regions = pd.DataFrame({
    'product': ['Laptop', 'Phone', 'Tablet', 'Monitor', 'Keyboard'],
    'region': ['North', 'South', 'North', 'South', 'North']
})
# Combine them with a shared key
combined = pd.merge(df, regions, on='product', how='inner')
combined
# Relate the joined data further
combined.groupby('region')['revenue'].sum()
```

- **Answer**: Combine them using a shared key. Example: pd.merge(df1, df2, on='common_id', how='inner')
