## 5-1. Conditional Tests

Ten tests, five True, five False.

```python
# Test variable
fruit = "pineapple"
print("Test 1: Is fruit == 'pineapple'? I predict True.")
print(fruit == "pineapple")

print("\nTest 2: Is fruit == 'banana'? I predict False.")
print(fruit == "banana")

# Number tests
number = 67
print("\nTest 3: Is number > 40? I predict True.")
print(number > 40)
print("\nTest 4: Is number < 40? I predict False.")
print(number < 40)

# Name tests
name = "James"
print("\nTest 5: Is name.lower() == 'james'? I predict True.")
print(name.lower() == "james")
print("\nTest 6: Is name == 'Bob'? I predict False.")
print(name == "Bob")

# Age tests
age = 30
print("\nTest 7: Is age >= 18? I predict True.")
print(age >= 18)
print("\nTest 8: Is age < 18? I predict False.")
print(age < 18)

# List tests
colors = ["red", "blue", "green"]
print("\nTest 9: Is 'red' in colors? I predict True.")
print("red" in colors)
print("\nTest 10: Is 'yellow' in colors? I predict False.")
print("yellow" in colors)
```

## 5-2. More Conditional Tests

More comparisons.

```python
# String equality
print("String tests:")
print("hello" == "hello")  # T
print("hello" != "world")  # T
print("hello" == "Hello")  # F

# lower() method
print("\nlower() tests:")
print("HELLO".lower() == "hello")  # T
print("Hello".lower() == "HELLO")  # F

# Numerical
print("\nNumerical tests:")
x = 10
print(x == 10)    # T
print(x != 5)     # T
print(x > 5)      # T
print(x < 5)      # F
print(x >= 10)    # T
print(x <= 5)     # F

# and/or
print("\nand/or tests:")
print((x > 5) and (x < 20))    # T
print((x > 5) and (x < 8))     # F
print((x > 15) or (x < 20))    # T
print((x > 15) or (x < 5))     # F

# in list
fruits = ["apple", "banana", "cherry"]
print("\n'in' tests:")
print("apple" in fruits)       # T
print("grape" in fruits)       # F

# not in list
print("\n'not in' tests:")
print("grape" not in fruits)   # T
print("apple" not in fruits)   # F
```

## 5-4. Alien Colors #2

if-else chain for alien.

```python
# Runs if block
alien_color = "green"

if alien_color == "green":
	print("You earned 5 points for shooting the alien!")
else:
	print("You earned 10 points!")

print("---")

# Runs else block
alien_color = "red"

if alien_color == "green":
	print("You earned 5 points for shooting the alien!")
else:
	print("You earned 10 points!")
```

## 5-5. Alien Colors #3

```python
# Green
alien_color = "green"

if alien_color == "green":
	print("You earned 5 points!")
elif alien_color == "yellow":
	print("You earned 10 points!")
else:
	print("You earned 15 points!")

print("---")

# Yellow
alien_color = "yellow"

if alien_color == "green":
	print("You earned 5 points!")
elif alien_color == "yellow":
	print("You earned 10 points!")
else:
	print("You earned 15 points!")

print("---")

# Red
alien_color = "red"

if alien_color == "green":
	print("You earned 5 points!")
elif alien_color == "yellow":
	print("You earned 10 points!")
else:
	print("You earned 15 points!")
```

## 5-8. Hello Admin

Greet users, special for admin.

```python
# Users list
usernames = ["admin", "Ashok", "James", "Nitesh", "Noman", "Fatima", "Ayesha", "Shamikh"]

# Greet each
for username in usernames:
	if username == "admin":
		print("Hello admin, would you like to see a status report?")
	else:
		print(f"Hello {username}, thank you for logging in again.")
```

## 5-10. Checking Usernames

Case-insensitive check.

```python
# Current users
current_users = ["Boss", "Rigby", "Th3vious", "Alice", "Bob", "Charlie", "Diana", "Eve"]
# New users
new_users = ["boss", "RIGBY", "Frank", "Grace", "Heidi"]

# Lowercase
current_users_lower = [user.lower() for user in current_users]

# Check each user
for new_user in new_users:
	if new_user.lower() in current_users_lower:
		print(f"Sorry, '{new_user}' is already taken. Please enter a new username.")
	else:
		print(f"'{new_user}' is available!")
```

## 5-11. Ordinal Numbers

1st, 2nd, 3rd, etc.

```python
# Numbers 1-9
numbers = list(range(1, 10))

# Print ordinals
for number in numbers:
	if number == 1:
		print(f"{number}st")
	elif number == 2:
		print(f"{number}nd")
	elif number == 3:
		print(f"{number}rd")
	else:
		print(f"{number}th")
```
