## 4-1. Pizzas

## 4-8. Cubes

First 10 cubes.

```python
# Calculate cubes
cubes = []
for number in range(1, 11):
	cube = number ** 3
	cubes.append(cube)
	print(cube)
```

## 4-9. Cube Comprehension

List comprehension cubes.

```python
# Cube comprehension
cubes = [number ** 3 for number in range(1, 11)]
print(cubes)
```

## 4-10. Slices

First, middle, last three items.

```python
# Cubes list
cubes = [number ** 3 for number in range(1, 11)]

# First three
print("The first three items in the list are:")
print(cubes[:3])

# Middle three
print("\nThree items from the middle of the list are:")
middle_start = len(cubes) // 2 - 1
print(cubes[middle_start:middle_start + 3])

# Last three
print("\nThe last three items in the list are:")
print(cubes[-3:])
```

## 4-11. My Pizzas, Your Pizzas

Copy list and add different items.

```python
# My pizzas
my_pizzas = ["Siciliana", "capricciosa", "diavola"]
# Copy for friend
friend_pizzas = my_pizzas[:]

# Add different
my_pizzas.append("meat lovers")
friend_pizzas.append("veggie")

# Print mine
print("My favorite pizzas are:")
for pizza in my_pizzas:
	print(f"  {pizza}")

# Print friend's
print("\nMy friend's favorite pizzas are:")
for pizza in friend_pizzas:
	print(f"  {pizza}")
```

## 4-12. More Loops

Two for loops for foods.

```python
# My foods
my_foods = ["pizza", "tacos", "cereal"]
# Copy for friend
friend_foods = my_foods[:]

# Add different
my_foods.append("cannoli")
friend_foods.append("ice cream")

# Print mine
print("My favorite foods are:")
for food in my_foods:
	print(f"  {food}")

# Print friend's
print("\nMy friend's favorite foods are:")
for food in friend_foods:
	print(f"  {food}")
```

## 4-13. Buffet

Tuple menu that can't change.

```python
# Original menu
menu = ("spaghetti", "meatball", "hero", "macaronni salad", "coffee")

# Print menu
print("Original menu:")
for food in menu:
	print(f"  {food}")

# Try to modify
print("\nTrying to modify the tuple...")
try:
	menu[0] = "pizza"
except TypeError as e:
	print(f"Error: {e}")

# Rewrite tuple
print("\nRevised menu:")
menu = ("pizza", "meatball", "hero", "macaronni salad", "cake")
for food in menu:
	print(f"  {food}")
```
