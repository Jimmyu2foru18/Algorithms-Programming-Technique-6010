## 2-4. Name Cases

Print name in lowercase, uppercase, and title case.

```python
# Store name
name = "James McGuigan"
# Show cases
print(name.lower())
print(name.upper())
print(name.title())
```

## 2-7. Stripping Names

Show whitespace stripping functions.

```python
# Name with whitespace
name = "\t\n  Guido van Rossum  \n\t"
# Show original
print("Original:")
print(repr(name))
# Strip left
print("\nlstrip():")
print(repr(name.lstrip()))
# Strip right
print("\nrstrip():")
print(repr(name.rstrip()))
# Strip both
print("\nstrip():")
print(repr(name.strip()))
```
