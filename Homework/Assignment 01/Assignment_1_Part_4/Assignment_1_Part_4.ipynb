now write some code that takes an integer and updates its value to its current value plus 3. Run this code in python tutor-- what happens-- does it create a new integer or is the same integer (with the same ID) in memory.
````python
x = 10
print("Original value:", x)
print("Original id:", id(x))

x = x + 3
print("Updated value:", x)
print("Updated id:", id(x))
print("Same ID?", id(x) == id(x))
````

- When we do `x = x + 3`, Python creates a new integer object in memory. The `id()` will change, which shows that integers are immutable.
- you cannot change the original integer object; you can only bind the variable name to a new object.

Use python tutor for memory visualization but Now with a List-- create a list and append a value to it, extend it by a given number of values. Did the ID change here? Does that mean lists are muteable or immuteable?

attach a picture of this from python tutor.

````python
my_list = [1, 2, 3]
print("Original list:", my_list)
print("Original id:", id(my_list))

my_list.append(4)
print("After append:", my_list)
print("ID after append:", id(my_list))

my_list.extend([5, 6])
print("After extend:", my_list)
print("ID after extend:", id(my_list))
print("Same ID throughout?", id(my_list) == id(my_list))
````

- The list's `id()` did not change after `append()` or `extend()`. This means lists are mutable.
- The same object in memory was modified in place rather than creating a new list object.
