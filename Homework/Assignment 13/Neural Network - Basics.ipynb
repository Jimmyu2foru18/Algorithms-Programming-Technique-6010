# Neural Net From Scratch
What is a  Neural Network?
- A neural network is a computational model inspired by the way biological neural networks in the human brain process information. 
- It consists of layers of interconnected nodes, or "**neurons**" that can learn to recognize patterns and make predictions based on input data. 
- Neural networks are a key component of machine learning and artificial intelligence, and they are used in various applications such as image and speech recognition, natural language processing, and autonomous systems.
### To understand the basics of what a neural networks, it's important to grasp the following concepts:
1. **Neurons**: The basic building blocks of a neural network, which receive input, process
    it, and produce an output. Each neuron applies a mathematical function to the input data and passes the result to the next layer of neurons.    

2. **Layers**: Neural networks are organized into layers, which can be categorized as input layers, hidden layers, and output layers. The input layer receives the raw data, hidden layers perform computations and feature extraction, and the output layer produces the final prediction or classification.

3. **Activation Functions**: These are mathematical functions applied to the output of each neuron to introduce non-linearity into the model. Common activation functions include the sigmoid, tanh, and ReLU (Rectified Linear Unit) functions. Activation functions help neural networks learn complex patterns in the data.

4. **Weights and Biases**: Each connection between neurons has an associated weight that determines the strength of the connection. Weights are adjusted during the training process to minimize the error in predictions. Biases are additional parameters that allow the model to shift the activation function, enabling better fitting of the data.

5. **Forward Propagation**: This is the process by which input data is passed through the network, layer by layer, to produce an output. During forward propagation, each neuron computes its output based on the inputs it receives and the weights and biases associated with its connections.

6. **Loss Function**: A loss function measures the difference between the predicted output and the actual target output. Common loss functions include mean squared error for regression tasks and cross-entropy loss for classification tasks. The goal of training a neural network is to minimize the loss function.

7. **Backpropagation**: This is the process of updating the weights and biases of the network based on the error calculated by the loss function. Backpropagation involves computing the gradient of the loss function with respect to each weight and bias, and then adjusting them in the opposite direction of the gradient to reduce the error.

8. **Training**: The process of feeding input data into the neural network, calculating the loss,   and updating the weights and biases through backpropagation. Training typically involves multiple iterations (epochs) over the dataset to improve the model's performance.

9. **Overfitting and Underfitting**: Overfitting occurs when a neural network learns the training data too well, capturing noise and details that do not generalize to new data. Underfitting occurs when the model is too simple to capture the underlying patterns in the data. Techniques such as regularization, dropout, and cross-validation are used to mitigate these issues.

10. **Hyperparameters**: These are parameters that are set before training the neural network, such as  the learning rate, number of layers, number of neurons per layer, and batch size. Hyperparameter tuning is an important step in optimizing the performance of a neural network.

### Example:
| $x_1$ | $x_2$ | $y$ |
| --- | --- | --- |
| 0 | 0 |  |
| 1 | 1 |  |
| 0 | 1 |  |
| 1 | 0 |  |

```python
# import libraries
import numpy as np

# Data: XOR problem
# 0 if both inputs are the same
# 1 if they are different
X = np.array([[0, 0], [1, 1], [0, 1], [1, 0]], dtype=float)
Y = np.array([[0], [0], [1], [1]], dtype=float)

# Network: 2 inputs, 2 hidden, 1 output
rng = np.random.default_rng(1) # random number generator
W1 = rng.random((2, 2)) # weights from input to hidden layer
W2 = rng.random((2, 1)) # weights from hidden to output layer
b1 = rng.random((2,)) # bias for hidden layer
b2 = rng.random((1,)) # bias for output layer


# Activation function: relu
def relu(x):
	return np.maximum(0.0, x)


# Activation function: sigmoid
def sigmoid(x):
	return 1.0 / (1.0 + np.exp(-x))


# Training loop
lr = 0.1 # learning rate
print("step  loss")

for step in range(1, 1501):
	#make predictions

	# forward pass (@ = matrix multiplication)
	h = relu(X @ W1 + b1) #hidden layer activations
	y_hat = sigmoid(h @ W2 + b2) #output layer activations

	# measure prediction error
	eps = 1e-12
	y_clipped = np.clip(y_hat, eps, 1.0 - eps)
	loss = -np.mean(Y * np.log(y_clipped) + (1.0 - Y) * np.log(1.0 - y_clipped))

	# backward pass
	d2 = (y_hat - Y) / len(X) # derivative of loss w.r.t. output layer
	d1 = (d2 @ W2.T) * (h > 0)

	# Update: a small step in the direction of the negative gradient
	W2 -= lr * (h.T @ d2)
	b2 -= lr * d2.sum(axis=0)
	W1 -= lr * (X.T @ d1)
	b1 -= lr * d1.sum(axis=0)

	if step == 1 or step % 150 == 0:
		print(f"{step:4d} {loss: .3f}")

print(f"\n{y_hat.round(2)}") # results after training
```

### Activation Functions

- **ReLU**: A non-linear activation function that outputs the input directly when it is positive and outputs zero otherwise. 
    - It is defined as $f(x) = \max(0, x)$. 
    - Deep learning models use ReLU widely because it reduces the vanishing gradient problem and speeds up neural network training.

- **Sigmoid**: A non-linear activation function that maps any real-valued number into the range $(0, 1)$. 
    - It is defined as $f(x) = \frac{1}{1 + \exp(-x)}$. 
    - You often see sigmoid in binary classification and in output layers that produce probabilities.

- **Tanh**: A non-linear activation function that maps any real-valued number into the range $(-1, 1)$. 
    - It is defined as $f(x) = \frac{\exp(x) - \exp(-x)}{\exp(x) + \exp(-x)}$. 
    - Networks often use tanh in hidden layers to center outputs around zero, which helps gradient flow during learning.
- **Linear**: A linear activation function that returns the input unchanged. 
    - It is defined as $f(x) = x$. 
    - You typically use a linear activation in the output layer for regression tasks.
- **Softmax**: A non-linear activation function that converts a vector of real-valued numbers into a probability distribution. 
    - It is defined as $f(x_i) = \frac{\exp(x_i)}{\sum_{j} \exp(x_j)}$ across all classes $j$. 
    - Models commonly use softmax in the output layer for multi-class classification so each class receives a probability that sums to 1 across the vector.
