## Read in the Fremont.csv and answer the following

### Q1. Can you add a datetime column?

```python
import pandas as pd

fremont = pd.read_csv('Datasets/Fremont.csv', parse_dates=['Date'])
fremont.rename(columns={'Date': 'datetime'}, inplace=True)
fremont.set_index('datetime', inplace=True)

fremont.head()
```

### Q2. Can you make another pandas data frame that adds up the totals for each day

```python
daily = fremont.resample('D').sum()
daily.head()
```

### Q3. Let's say a sensor "Wins" the hour if it gets the most people going through it. can you count how many wins each sensor got according to this data

```python
print(fremont.columns)
```

```python
sensors = ['Fremont Bridge Sidewalks, south of N 34th St Total',
           'Fremont Bridge Sidewalks, south of N 34th St Cyclist West Sidewalk',
           'Fremont Bridge Sidewalks, south of N 34th St Cyclist East Sidewalk']

fremont_numeric = fremont[sensors].replace({',': ''}, regex=True).apply(pd.to_numeric, errors='coerce')
fremont_clean = fremont_numeric.dropna(how='all')

winner = fremont_clean.idxmax(axis=1)
wins = winner.value_counts()
wins
```

## Questions 5-8 Read in the Seattle.csv and answer the following

### Q5. Create a joined dataset  using the days of the weather sheet

```python
seattle['date'] = seattle['datetime'].dt.date
daily.index = daily.index.date

joined = seattle.merge(daily, left_on='date', right_index=True, how='inner')
joined.head()
```

### Q6. What was the total count of bikers on days where there was precipitation vs days where there wasn't precipitation

```python
rain = joined[joined['precip'] > 0]
no_rain = joined[joined['precip'] == 0]

rain_total = rain[sensors[0]].sum()
no_rain_total = no_rain[sensors[0]].sum()

rain_total, no_rain_total
```

### Q7. how many bikers are there if the max temp was above 90, 80-90, 70-80, 60-70, 50-60, below 60

```python
def temp_bin(t):
    if t > 90:
        return 'above 90'
    elif t >= 80:
        return '80-90'
    elif t >= 70:
        return '70-80'
    elif t >= 60:
        return '60-70'
    elif t >= 50:
        return '50-60'
    else:
        return 'below 60'

joined['temp_bin'] = joined['tempmax'].apply(temp_bin)
joined.groupby('temp_bin')[sensors[0]].sum()
```

### Q8. Categorize the windspeeds as low medium and high. Now determine how many bikers there were at each wind speed.

```python
def wind_cat(w):
    if w < 5:
        return 'low'
    elif w <= 15:
        return 'medium'
    else:
        return 'high'

joined['wind_cat'] = joined['windspeed'].apply(wind_cat)
joined.groupby('wind_cat')[sensors[0]].sum()
```
