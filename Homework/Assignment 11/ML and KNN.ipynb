# ML and KNN Assignment (Iris Dataset)

## Imports

```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
```

## Task 1) Read data from seaborn/Iris CSV

```python
iris = sns.load_dataset('iris')
iris.to_csv('iris.csv', index=False)
df = pd.read_csv('iris.csv')
df.head()
```

## Task 2) X matrix of features and y vector of labels

```python
X = df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']].values
y = df['species'].values
print('X shape:', X.shape)
print('y shape:', y.shape)
print('Classes:', np.unique(y))
```

## Task 3) train_test_split

```python
Xtrain, Xtest, ytrain, ytest = train_test_split(X, y, test_size=0.25, random_state=42)
print(Xtrain.shape, Xtest.shape, ytrain.shape, ytest.shape)
```

## Tasks 4-9) KNN from scratch

```python
def knn_predict(Xtrain, ytrain, Xtest, k=3):
    predictions = []
    for p in range(len(Xtest)):
        test_row = Xtest[p]
        distances = np.sqrt(np.sum((Xtrain - test_row)**2, axis=1))
        k_idx = np.argpartition(distances, k)[:k]
        k_labels = ytrain[k_idx]
        unique, counts = np.unique(k_labels, return_counts=True)
        pred = unique[counts.argmax()]
        predictions.append(pred)
    return np.array(predictions)

y_pred = knn_predict(Xtrain, ytrain, Xtest, k=3)
print(y_pred)
```

```python
correct = np.sum(y_pred == ytest)
total = len(ytest)
accuracy = correct/total
print('k=3:', correct, '/', total, 'correct, accuracy=', accuracy)
for k_try in [1,3,5,7,9]:
    yp = knn_predict(Xtrain,ytrain,Xtest,k=k_try)
    acc = np.sum(yp==ytest)/len(ytest)
    print('k='+str(k_try)+': accuracy='+str(acc))
```
