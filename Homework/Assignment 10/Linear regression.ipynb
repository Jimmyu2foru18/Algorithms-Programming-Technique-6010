# Idea Behind Linear Regression

Linear regression is used when we have a set of features and we want to predict a **continuous variable target**. Examples of continuous variable targets would be Housing Price, Person's weight, Gas Prices, etc.

Note that this differs from **Classification**, where the task is to predict a category given the values in feature space (admission or not admission, cancer or benign, spam or good mail).

In linear regression our predictive equation will have the form:

**Y = b₀ + b₁·X₁ + b₂·X₂ + …**

- **Y** is what we are trying to predict.
- **X₁, X₂, …** are our features.
- **b₀** is the y-intercept.
- **b₁, b₂, …** give us how a unit change in feature X₁ will be reflected in Y. In other words, they tell us **how sensitive Y is to each of these features**.

### Our basic approach to linear regression is as follows:

1. **Identify our Target value** (aka the value we want to make a prediction on) in the data.
2. Use the functions `head` and `sns.pairplot` to get an idea of the data.
3. **Calculate the Correlation statistics** from our DataFrame and use `sns.heatmap` with the correlation to quickly see what data is correlated to our target value and possibly cross-correlated to one another.
4. From the correlation, **pick the features that are correlated with the target without being correlated strongly to each other**. (There is a more in-depth explanation of why this is the case here:
   - [Is there an intuitive explanation why multicollinearity is a problem in linear regression? (StackExchange)](https://stats.stackexchange.com/questions/1149/is-there-an-intuitive-explanation-why-multicollinearity-is-a-problem-in-linear-r/1150#1150)
   - [Features, correlations, data leakage, confounded features, and other things… (Towards Data Science)](https://towardsdatascience.com/features-correlations-data-leakage-confounded-features-and-other-things-that-can-make-your-deep-771bcaf84f9f)

   > This idea of when to not include a co-correlated feature might be something to bring up to your Statistics Professor.
5. **Import** from scikit-learn: `train_test_split` and `LinearRegression`. (`.2` or `.4` test size is typical.)
6. Create a new `LinearRegression` object and use it to **fit your training data**.
7. You can now see the `.intercept_` and `.coef_` of the model (i.e., if `lr` is your model object, you can say `lr.coef_` to see the coefficients (b₁, b₂, …) and `lr.intercept_` to see b₀).
8. **Use your model to predict.**
