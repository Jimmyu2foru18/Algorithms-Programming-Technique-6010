```python
from pathlib import Path
import csv
import numpy as np
```

Which Cereal has the most fiber
get it into a matrix 

or find average of fiber in each of the four quartiles 

```python
script_dir = Path.cwd()
file_path = script_dir / 'cereal.csv'

with open(file_path, 'r') as f:
  reader = csv.reader(f)
  data = list(reader)

data_array = np.array(data)

header = data_array[0]
rows = data_array[1:]

name_idx = list(header).index('name')
fiber_idx = list(header).index('fiber')
```

```python
fibers = rows[:, fiber_idx].astype(float)
max_fiber_idx = np.argmax(fibers)
print('Most fiber cereal:', rows[max_fiber_idx, name_idx])
```

```python
sorted_fibers = np.sort(fibers)
quartiles = np.array_split(sorted_fibers, 4)

for index, group in enumerate(quartiles, start=1):
  average_fiber = np.mean(group)
  print(f'Quartile {index} average fiber: {average_fiber:.2f}')
```
